## Ejercicio 1: Iterar sobre lista
Crea una lista con las fechas y horas de clase desde hoy (14 de septiembre de 2026) hasta la última clase (4 de diciembre de 2026). Ten en cuenta que los festivos no serán horas de clase.

```python
from datetime import datetime, timedelta, date

inicio = datetime(2026, 9, 14, 15, 15, 0, 0) # Año, mes, día, horas, minutos, segundos y microsegundos
final = datetime(2026, 12, 8, 15, 15, 0, 0) # Año, mes, día, horas, minutos, segundos y microsegundos

clases = [inicio]

while clases[-1] <= final:
    clases.append(clases[-1] + timedelta(weeks = 1))

print(len(clases))
```
